```python
import torch
import clip
from PIL import Image
```

```python
def calculate_single_clip_score(image_path, prompt):

    device = "cuda" if torch.cuda.is_available() else "cpu"
    model, preprocess = clip.load("ViT-B/32", device=device)


    text = clip.tokenize([prompt]).to(device)


    try:
        image = preprocess(Image.open(image_path)).unsqueeze(0).to(device)
    except Exception as e:
        print(f"Error loading image: {e}")
        return


    with torch.no_grad():
        logits_per_image, _ = model(image, text)
        score = logits_per_image.item()

    print(f"--- Evaluation Result ---")
    print(f"Image Path: {image_path}")
    print(f"Prompt Used: {prompt}")
    print(f"CLIP Score: {score:.2f}")

my_image_path = "/content/images/WhatsApp Image 2026-02-18 at 1.09.08 PM.jpeg"


my_prompt = "When the children opened a “serious detective agency” to find who stole the last cookieو they spent an hour interrogating goldfish."

calculate_single_clip_score(my_image_path, my_prompt)
```

```python
import torch
import clip
from PIL import Image
import os
import numpy as np

def calculate_average_clip(folder_path, prompt):

    device = "cuda" if torch.cuda.is_available() else "cpu"
    model, preprocess = clip.load("ViT-B/32", device=device)

    text = clip.tokenize([prompt]).to(device)

    clip_scores = []
    valid_extensions = ('.png', '.jpg', '.jpeg', '.bmp', '.webp')

    print(f"Calculating CLIP scores for images in: {folder_path}")


    for filename in os.listdir(folder_path):
        if filename.lower().endswith(valid_extensions):
            img_path = os.path.join(folder_path, filename)
            try:

                image = preprocess(Image.open(img_path)).unsqueeze(0).to(device)

                with torch.no_grad():

                    image_features = model.encode_image(image)
                    text_features = model.encode_text(text)


                    image_features /= image_features.norm(dim=-1, keepdim=True)
                    text_features /= text_features.norm(dim=-1, keepdim=True)

                    similarity = (image_features @ text_features.T).item()
                    clip_scores.append(similarity)

            except Exception as e:
                print(f"Error processing {filename}: {e}")


    if not clip_scores:
        return 0

    avg_score = np.mean(clip_scores)
    return avg_score


my_folder = r"/content/images"
my_prompt = "When the children opened a “serious detective agency” to find who stole the last cookie, they spent an hour interrogating the class goldfish "

average_result = calculate_average_clip(my_folder, my_prompt)

print("-" * 30)
print(f"Average CLIP Score: {average_result:.4f}")
print("-" * 30)
```

```python
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer
import logging
from transformers import logging as transformers_logging

transformers_logging.set_verbosity_error()

def calculate_professional_ppl(text):

    model_id = "gpt2-xl"
    tokenizer = GPT2Tokenizer.from_pretrained(model_id)
    model = GPT2LMHeadModel.from_pretrained(model_id)


    clean_text = " ".join(text.split())
    clean_text = clean_text.replace('“', '"').replace('”', '"').replace('’', "'")

    inputs = tokenizer(clean_text, return_tensors="pt")
    input_ids = inputs["input_ids"]

    with torch.no_grad():
        outputs = model(input_ids, labels=input_ids)
        loss = outputs.loss

    return torch.exp(loss).item()

my_story = """When the children opened a "serious detective agency" to find who stole the last cookie, they spent an hour interrogating the class goldfish. They finally solved the mystery when crumbs on Tommy's shirt testified louder than any suspect ever could."""

ppl = calculate_professional_ppl(my_story)
print(f"Professional Perplexity Score: {ppl:.2f}")
```
